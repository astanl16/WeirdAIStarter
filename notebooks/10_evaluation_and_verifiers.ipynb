## 1. Why Evaluate?

Training a model is only half of AI engineering. We also need a reliable way to measure whether the model is producing useful results.

Chapter 3 introduces the idea of a **verifier**:

```
Prompt
   ↓
LLM
   ↓
Generated Response
   ↓
Extract Final Result
   ↓
Normalize
   ↓
Verify
   ↓
Score
```

Some domains (math, code) allow exact verification. Others (creative writing) require heuristic scoring.


````python
# Example model response

response = '''
Reasoning:
The cat sat on the mat.

Final Answer:
42
'''

print(response)
````

## 2. Extracting Information

Most LLM responses contain extra reasoning that is not needed for evaluation.

A verifier usually extracts only the important portion of the response.


````python
def extract_after_marker(text, marker):
    # Simple example implementation
    if marker not in text:
        return text.strip()
    return text.split(marker,1)[1].strip()

answer = extract_after_marker(response, "Final Answer:")
print(answer)
````

## 3. Normalization

Two answers may be logically identical while looking different.

Examples:

- `" Hello "` vs `"Hello"`
- extra blank lines
- inconsistent capitalization

Normalization converts data into a consistent format before comparison.


````python
examples = [
    "  Hello  ",
    "\nHello\n",
    "HELLO"
]

normalized = [x.strip().lower() for x in examples]

for before, after in zip(examples, normalized):
    print(f"{before!r} -> {after!r}")
````

## 4. Exact vs. Heuristic Evaluation

| Exact | Heuristic |
|---|---|
| 2 + 2 = 4 | Does this lyric rhyme? |
| SQL query returns expected rows | Is the parody funny? |
| Unit tests pass | Does the song keep a consistent meter? |

Creative AI systems often combine several heuristic scores into one overall score.


````python
scores = {
    "rhyme": 0.82,
    "syllables": 0.91,
    "structure": 1.00
}

overall = sum(scores.values()) / len(scores)

print(scores)
print(f"Overall score: {overall:.2f}")
````
