```python
from pathlib import Path

text = Path("../data/processed/lyrics_sample.txt").read_text(
    encoding="utf-8"
)

print(text[:1000])
```

```python
lines = text.splitlines()

non_empty_lines = [
    line.strip()
    for line in lines
    if line.strip()
]

average_length = sum(len(line) for line in non_empty_lines) / len(non_empty_lines)

print(f"Average line length: {average_length:.2f}")
```

```python
from weird_ai.tokenizer import SimpleCharacterTokenizer

tokenizer = SimpleCharacterTokenizer(text)

print(tokenizer.chars)
print(len(tokenizer.chars))
```

```python
encoded = tokenizer.encode("hello class")

print(encoded)
```

```python
decoded = tokenizer.decode(encoded)

print(decoded)
```
